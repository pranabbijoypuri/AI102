#### Create `Face API` resource in Azure portal

⚙️ Install the following package
- >poetry add azure-ai-vision-face

```python
import os
from dotenv import load_dotenv
from azure.core.credentials import AzureKeyCredential

from azure.ai.vision.face import FaceClient
from azure.ai.vision.face.models import FaceDetectionModel, FaceRecognitionModel, FaceAttributeTypeDetection01
```

```python
load_dotenv()
ai_endpoint = os.getenv(key="AI_SERVICE_FACE_API_ENDPOINT")
ai_key = os.getenv(key="AI_SERVICE_FACE_API_KEY")
```

```python
face_client = FaceClient(endpoint=ai_endpoint,
                         credential=AzureKeyCredential(key=ai_key)
                        )
```

```python
image_file_path = "./images/faces.jpg"
image_data = open(file=image_file_path, mode='rb')

result = face_client.detect(image_content=image_data.read(),
                            detection_model=FaceDetectionModel.DETECTION01,
                            recognition_model=FaceRecognitionModel.RECOGNITION01,
                            return_face_id=False,
                            return_face_attributes=[
                                FaceAttributeTypeDetection01.HEAD_POSE,
                                FaceAttributeTypeDetection01.OCCLUSION,
                                FaceAttributeTypeDetection01.ACCESSORIES                                         
                            ]
                            )
```

```python
i = 1
for detected_face in result:
    print(f"😐 Face {i}:")
    print(f"Face Attribute(Head Pose): {detected_face.face_attributes.head_pose}")
    print(f"Face Attribute(Accessories): {detected_face.face_attributes.accessories}")
```
