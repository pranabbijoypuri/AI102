⚙️ Install the following package
- >poetry add azure-ai-vision-imageanalysis

```python
from azure.ai.vision.imageanalysis import ImageAnalysisClient
from azure.ai.vision.imageanalysis.models import VisualFeatures
from azure.core.credentials import AzureKeyCredential
import os
from dotenv import load_dotenv
```

```python
load_dotenv()
ai_endpoint = os.getenv(key="AI_SERVICE_COMPUTER_VISION_ENDPOINT")
ai_key = os.getenv(key="AI_SERVICE_COMPUTER_VISION_KEY")
```

```python
ocr_client = ImageAnalysisClient(endpoint=ai_endpoint,
                                 credential=AzureKeyCredential(key=ai_key)
                                )
```

```python
image_path = "./images/Business-card.jpg"
image_data = open(file=image_path, mode='rb')
# Read the data from file
result = ocr_client.analyze(image_data=image_data,
                            visual_features=[VisualFeatures.READ]
                        )
```

```python
result.read.blocks[0].lines
```

```python
i=1
for single_line in result.read.blocks[0].lines:
    print(f"🗣️ Line {i}: {single_line.text}")
    for sinlge_word in single_line.words:
        print(f"    ✨{sinlge_word.text}  (Confidence: {sinlge_word.confidence*100})")
    i+=1
```
