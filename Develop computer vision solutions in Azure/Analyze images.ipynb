```python
from azure.core.credentials import AzureKeyCredential
from azure.ai.vision.imageanalysis import ImageAnalysisClient
from azure.ai.vision.imageanalysis.models import VisualFeatures

from dotenv import load_dotenv
import os
```

```python

load_dotenv()
ai_endpoint = os.getenv(key="AI_SERVICE_COMPUTER_VISION_ENDPOINT")
ai_key = os.getenv(key="AI_SERVICE_COMPUTER_VISION_KEY")
```

```python
# Authenticate Azure AI Vision client
cv_client = ImageAnalysisClient(endpoint=ai_endpoint,
                                credential=AzureKeyCredential(key=ai_key)
                            )
```

🧠 Analuze image to get the features.

```python
# Open image as binary object
image_path = "./images/street.jpg"
with open(image_path, "rb") as f:
    image_data = f.read()


# Analyze image 
result = cv_client.analyze(
    image_data=image_data,
    visual_features=[
        VisualFeatures.CAPTION,
        VisualFeatures.DENSE_CAPTIONS,
        VisualFeatures.TAGS,
        VisualFeatures.OBJECTS,
        VisualFeatures.PEOPLE
    ]
)
```

```python
# Get the caption
print(f"CAPTION: {result.caption.text}")
print(f"Caption Confidence: {result.caption.confidence}")
```

```python
# Get the dense caption
i = 1
for dense_cap in result.dense_captions.list:
    print(f"CAPTION {i}: {dense_cap.text}")
    print(f"CAPTION AREA {i}: {dense_cap.bounding_box}")
    print(f"CAPTION Confidence {i}: {dense_cap.confidence}")
    i+=1

```
